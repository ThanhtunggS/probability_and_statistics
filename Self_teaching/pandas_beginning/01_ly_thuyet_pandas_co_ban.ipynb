## Phần 1: Khởi động - Import thư viện
Để sử dụng Pandas, ta dùng câu lệnh `import pandas as pd`. 
Đồng thời, ta import thêm hàm `display` từ `IPython.display` để hiển thị bảng dữ liệu đẹp mắt dạng bảng kẻ ô.


```python
import pandas as pd
from IPython.display import display

print("Phiên bản Pandas hiện tại:", pd.__version__)
```

---
## Phần 2: Hai cấu trúc dữ liệu cốt lõi trong Pandas
Trong Pandas, bạn sẽ làm việc với 2 kiểu cấu trúc chính:
- **Series:** Một danh sách dữ liệu 1 chiều (tương tự như 1 cột đơn lẻ).
- **DataFrame:** Bảng dữ liệu 2 chiều hoàn chỉnh, gồm nhiều hàng (rows) và nhiều cột (columns).

Hãy chạy ô code dưới đây để xem cách tạo một DataFrame đơn giản từ một `Dictionary`:


```python
# Tạo dữ liệu dạng Dictionary (Từ điển)
du_lieu_mau = {
    "ten": ["An", "Bình", "Cường"],
    "tuoi": [19, 20, 19],
    "diem": [8.5, 9.0, 7.5]
}

# Chuyển đổi thành DataFrame
df_vi_du = pd.DataFrame(du_lieu_mau)

print("Loại đối tượng df_vi_du:", type(df_vi_du))
print("Loại đối tượng của 1 cột df_vi_du['ten']:", type(df_vi_du['ten']))

# Hiển thị DataFrame dưới dạng bảng HTML
display(df_vi_du)
```

---
## Phần 3: Đọc dữ liệu từ file CSV bên ngoài (`pd.read_csv`)
Trong thực tế, ta không tự gõ tay dữ liệu vào code mà đọc từ các file dữ liệu (như `.csv`, `.xlsx`).
Hàm `pd.read_csv("đường_dẫn_tới_file.csv")` sẽ tự động đọc file và nạp toàn bộ vào một biến DataFrame trong bộ nhớ RAM.

File mẫu của chúng ta đang nằm tại: `du-lieu/sinh_vien.csv`.


```python
# Đọc dữ liệu từ thư mục du-lieu
df_sv = pd.read_csv("du-lieu/sinh_vien.csv")

# Hiển thị toàn bộ bảng dữ liệu
display(df_sv)
```

---
## Phần 4: Khám phá tổng quan bảng dữ liệu (Data Exploration)
Khi nhận một tập dữ liệu mới, có 5 lệnh cơ bản đầu tiên bạn luôn nên chạy:

1. `df.shape`: Trả về số hàng và số cột dạng tuple `(hàng, cột)`.
2. `df.head(n)`: Xem nhanh `n` hàng đầu tiên (mặc định là 5 hàng).
3. `df.tail(n)`: Xem nhanh `n` hàng cuối cùng.
4. `df.info()`: Xem tên các cột, số lượng giá trị không rỗng (`Non-Null Count`), và kiểu dữ liệu máy tính (`dtypes`).
5. `df.describe()`: Tóm tắt thống kê nhanh cho tất cả các cột kiểu số (trung bình, độ lệch chuẩn, min, max...).


```python
# 1. Kích thước bảng
print("Kích thước bảng (Số hàng, Số cột):", df_sv.shape)
print("Số hàng:", df_sv.shape[0])
print("Số cột:", df_sv.shape[1])
```

```python
# 2. Xem 3 hàng đầu tiên
print("--- 3 Hàng đầu tiên ---")
display(df_sv.head(3))

# 3. Xem 2 hàng cuối cùng
print("--- 2 Hàng cuối cùng ---")
display(df_sv.tail(2))
```

```python
# 4. Kiểm tra cấu trúc kỹ thuật dtypes và giá trị null
print("--- Thông tin cấu trúc dữ liệu ---")
df_sv.info()
```

```python
# 5. Tóm tắt thống kê các biến định lượng (số thực, số nguyên)
print("--- Bảng tóm tắt thống kê mô tả ---")
display(df_sv.describe())
```

---
## Phần 5: Trích xuất cột dữ liệu
- **Lấy 1 cột:** Dùng cú pháp `df["ten_cot"]` (kết quả trả về dạng `Series`).
- **Lấy nhiều cột:** Dùng cú pháp danh sách trong ngoặc vuông kép: `df[["cot_1", "cot_2", ...]]` (kết quả trả về dạng `DataFrame` con).


```python
# Lấy riêng cột điểm GPA
cot_gpa = df_sv["gpa"]
print("Kiểu dữ liệu khi lấy 1 cột:", type(cot_gpa))
print(cot_gpa)
```

```python
# Lấy một bảng con gồm 3 cột: Họ tên, Ngành học, GPA
bang_con = df_sv[["ho_ten", "nganh_hoc", "gpa"]]
display(bang_con)
```

---
## Phần 6: Lọc dữ liệu theo điều kiện (Filtering Rows)
Đây là một trong những thao tác quan trọng nhất trong phân tích dữ liệu.  
Cú pháp cơ bản: `df[điều_kiện]`

- Điều kiện so sánh số: `df["gpa"] >= 3.5`
- Điều kiện so sánh chuỗi: `df["nganh_hoc"] == "CNTT"`
- Kết hợp nhiều điều kiện:
  - Phép VÀ (AND): `df[(điều_kiện_1) & (điều_kiện_2)]`
  - Phép HOẶC (OR): `df[(điều_kiện_1) | (điều_kiện_2)]`


```python
# Ví dụ 1: Lọc sinh viên có điểm GPA từ 3.5 trở lên
sv_gioi = df_sv[df_sv["gpa"] >= 3.5]

print(f"Có {len(sv_gioi)} sinh viên đạt GPA >= 3.5:")
display(sv_gioi[["ho_ten", "nganh_hoc", "gpa"]])
```

```python
# Ví dụ 2: Lọc sinh viên học ngành CNTT
sv_cntt = df_sv[df_sv["nganh_hoc"] == "CNTT"]
display(sv_cntt[["ho_ten", "nganh_hoc", "que_quan"]])
```

```python
# Ví dụ 3: Lọc sinh viên học CNTT VÀ có khoảng cách từ nhà đến trường > 4.0 km
sv_cntt_o_xa = df_sv[(df_sv["nganh_hoc"] == "CNTT") & (df_sv["khoang_cach_km"] > 4.0)]
display(sv_cntt_o_xa)
```

---
## Phần 7: Các phép thống kê cơ bản trên cột
Pandas hỗ trợ các hàm thống kê tính toán trực tiếp trên cột:
- `.mean()`: Tính giá trị trung bình cộng.
- `.median()`: Tìm số trung vị.
- `.max()` và `.min()`: Tìm giá trị lớn nhất, nhỏ nhất.
- `.sum()`: Tính tổng các giá trị.
- `.value_counts()`: Đếm số lần xuất hiện (tần số) của từng nhóm giá trị phân loại.


```python
# Tính toán các chỉ số thống kê trên cột gpa
gpa_mean = df_sv["gpa"].mean()
gpa_max = df_sv["gpa"].max()
gpa_min = df_sv["gpa"].min()

print(f"GPA trung bình cả lớp: {gpa_mean:.2f}")
print(f"GPA cao nhất: {gpa_max} | GPA thấp nhất: {gpa_min}")
```

```python
# Đếm số lượng sinh viên theo từng ngành học (Biến phân loại)
tan_so_nganh = df_sv["nganh_hoc"].value_counts()
print("Số lượng sinh viên theo từng ngành học:")
print(tan_so_nganh)
```

## Tóm tắt Cheat Sheet ghi nhớ nhanh
| Lệnh | Ý nghĩa |
| :--- | :--- |
| `pd.read_csv(path)` | Đọc file CSV thành DataFrame |
| `df.head(n)` | Xem `n` hàng đầu tiên |
| `df.shape` | Kích thước bảng (hàng, cột) |
| `df.info()` | Kiểm tra kiểu dữ liệu các cột |
| `df.describe()` | Bảng thống kê tóm tắt các cột số |
| `df["cot"]` | Lấy 1 cột đơn lẻ (Series) |
| `df[["c1", "c2"]]` | Trích xuất bảng con nhiều cột |
| `df[df["cot"] > x]` | Lọc các hàng thỏa mãn điều kiện |
| `df["cot"].mean()` | Tính giá trị trung bình |
| `df["cot"].value_counts()` | Đếm tần số xuất hiện của từng phân loại |
