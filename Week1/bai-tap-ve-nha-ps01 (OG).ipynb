### Bài 2: Tính toán các chỉ số thống kê mô tả cơ bản
1. Tính khoảng cách trung bình đến trường của các bạn (`mean`).

3. Lọc ra danh sách những bạn có khoảng cách từ nhà đến trường lớn hơn 3.0 km.

1. Đọc tệp CSV vào biến `df_loan`.
2. Kiểm tra số hàng và số cột của bảng dữ liệu (`df_loan.shape`).
3. In ra danh sách các cột và kiểu dữ liệu `dtypes` tương ứng.
4. Xác định đâu là biến định lượng liên tục, đâu là biến phân loại thứ bậc.

```python
import pandas as pd
df_loan = pd.read_csv('du-lieu/loan50.csv')
print(df_loan.shape)
print(df_loan.dtypes)
display(df_loan.head())
bien_dinh_luong_lien_tuc = df_loan.select_dtypes(include=['float64']).columns.tolist()
bien_dinh_luong_lien_tuc.extend(['annual_income', 'loan_amount', 'total_income'])
print("Biến định lượng liên tục:", list(set(bien_dinh_luong_lien_tuc)))
```
