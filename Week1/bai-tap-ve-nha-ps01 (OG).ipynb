### Bài 3: Khám phá tập dữ liệu thực tế `loan50.csv`
Tập dữ liệu `loan50.csv` gồm 50 khoản vay từ Lending Club được lưu tại thư mục `du-lieu/loan50.csv`.
1. Đọc tệp CSV vào biến `df_loan`.
2. Kiểm tra số hàng và số cột của bảng dữ liệu (`df_loan.shape`).
3. In ra danh sách các cột và kiểu dữ liệu `dtypes` tương ứng.
4. Xác định đâu là biến định lượng liên tục, đâu là biến phân loại thứ bậc.

```python
import os

# Đường dẫn tới file loan50.csv
csv_path = "../../du-lieu/loan50.csv"
if not os.path.exists(csv_path):
    csv_path = "../03-thuc-hanh-python/du-lieu/loan50.csv"

df_loan = pd.read_csv(csv_path)
print("Kích thước tập dữ liệu (hàng, cột):", df_loan.shape)
print("\n--- 5 hàng đầu tiên ---")
display(df_loan.head())

print("\n--- Kiểu dữ liệu kỹ thuật dtypes ---")
print(df_loan.dtypes)
```

### Bài 4: Chuyển đổi kiểu dữ liệu phân loại thứ bậc (`Categorical Ordinal`)
Cột `grade` (xếp hạng tín dụng) trong `df_loan` gồm các bậc từ `A` (an toàn nhất) đến `G` (rủi ro nhất).
Hãy chuyển đổi cột `grade` thành kiểu `category` có thứ tự (`ordered=True`) trong pandas.

```python
grades_order = ["A", "B", "C", "D", "E", "F", "G"]
df_loan["grade"] = pd.Categorical(df_loan["grade"], categories=grades_order, ordered=True)

print("Kiểu dữ liệu của grade sau khi chuyển:", df_loan["grade"].dtype)
print("Tần số từng bậc xếp hạng:")
print(df_loan["grade"].value_counts().sort_index())
```
